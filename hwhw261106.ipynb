### 2. loc vs iloc

1. `loc`으로 인덱스 10~15 행의 `sex`, `age` 열을 선택해 출력하세요.
2. `iloc`으로 같은 열(`sex`는 2번째, `age`는 3번째 위치)을 행 위치 10:15로 선택해 출력하세요.
3. 두 결과의 행 수를 각각 출력하고, 왜 다른지 주석으로 쓰세요.

```python
loc_result = titanic.loc[10:15, ["sex", "age"]]                         #loc는 끝 위치 15가 포함
print(loc_result)
print("loc 행 수:", len(loc_result),"-----loc는 끝 위치 15가 포함")

iloc_result = titanic.iloc[10:15, [2, 3]]                               #iloc는 끝 위치 15가 제외
print("\n",iloc_result)
print("iloc 행 수:", len(iloc_result),"-----iloc는 끝 위치 15가 제외")
```

### 5. 복합 조건

각각 조건을 만족하는 행 수를 출력하세요.

1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

```python
pclass3_male_survived = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["survived"] == 1)
pclass1_or_high_fare = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
not_alone = ~titanic["alone"]

print("3등석 남성 생존자:", pclass3_male_survived.sum())
print("1등석 또는 요금 100 이상:", pclass1_or_high_fare.sum())
print("혼자 탑승하지 않은 승객:", not_alone.sum())
```

### 6. isin · between

1. `isin`으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 `shape`를 출력하세요.
2. `between`으로 나이 20~30세(양 끝 포함) 승객을 골라 `shape`와 나이의 min·max를 출력하세요.
3. 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.

```python
cherbourg_or_queenstown = titanic[titanic["embark_town"].isin(["Cherbourg", "Queenstown"])]
print("Cherbourg 또는 Queenstown:", *cherbourg_or_queenstown.shape)

age_20_30 = titanic[titanic["age"].between(20, 30)]
print("\n20~30세:", *age_20_30.shape)
print(f"\n나이\n최솟값 : {int(age_20_30["age"].min())} \n최댓값 : {int(age_20_30["age"].max())}")

age_20_30_exclusive = titanic[titanic["age"].between(20, 30, inclusive="neither")]
print("\n20~30세:", *age_20_30_exclusive.shape,"(양 끝 미포함)")
```

### 7. loc[조건, 열]

12살 이하 **그리고** 3등석 승객의 `age`, `sex`, `survived` 열만 `kids` 변수에 담고, 행 수와 생존률(`survived` 평균, 소수 둘째 자리)을 출력하세요.

```python
kids = titanic.loc[(titanic["age"] <= 12) & (titanic["pclass"] == 3), ["age", "sex", "survived"]]
print("행 수:", len(kids))
print("생존률:", round(kids["survived"].mean(), 2))
```

### 8. query

1. 변수 `min_fare = 50`을 만들고, `query`로 Cherbourg 탑승 **그리고** 요금이 `min_fare` 이상인 승객의 `shape`를 출력하세요.
2. 5-1번 조건을 `query`로 다시 써서 행 수가 같은지 확인하세요.

```python
min_fare = 50
cherbourg_expensive = titanic.query("embark_town == 'Cherbourg' and fare >= @min_fare")
print("Cherbourg 탑승, 요금 50 이상:", *cherbourg_expensive.shape)

query_third_male_survived = titanic.query("pclass == 3 and sex == 'male' and survived == 1")
print(f"5-1 조건 행 수: (기존){int(pclass3_male_survived.sum())}, (query){len(query_third_male_survived)}")
```

### 9. 정렬

1. `pclass` 오름차순, 같은 등급 안에서는 `fare` 내림차순으로 정렬해 `pclass`, `fare`, `sex`, `age` 열의 상위 3행을 출력하세요.
2. 요금 내림차순으로 정렬한 결과를 변수에 담고, `iloc[0]`과 `loc[0]`의 `fare` 값을 각각 출력하세요. 두 값이 다른 이유를 주석으로 쓰세요.

```python
sorted_by_class_fare = titanic.sort_values(["pclass", "fare"], ascending=[True, False])
print(sorted_by_class_fare[["pclass", "fare", "sex", "age"]].head(3))

sorted_by_fare = titanic.sort_values("fare", ascending=False)
print("\niloc[0] fare:", sorted_by_fare.iloc[0]["fare"])
print("loc[0] fare:", sorted_by_fare.loc[0, "fare"])
# iloc[0]은 정렬 후 첫 행, loc[0]은 원래 인덱스 라벨이 0인 행을 선택합니다.
```

### 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
male_age_20_40_first_second = titanic.loc[
    (titanic["sex"] == "male")
    & titanic["age"].between(20, 40)
    & titanic["pclass"].isin([1, 2]),
    ["pclass", "age", "fare", "survived"],
]
print("1·2등석 행 수:", len(male_age_20_40_first_second))
print("1·2등석 생존률:", round(male_age_20_40_first_second["survived"].mean(), 2),"%")
print("\n요금 상위 3명")
print(male_age_20_40_first_second.sort_values("fare", ascending=False).head(3))

query_first_second = titanic.query("sex == 'male' and age >= 20 and age <= 40 and pclass in [1, 2]")
print("\nquery 행 수:", len(query_first_second))

male_age_20_40_third = titanic.loc[
    (titanic["sex"] == "male")
    & titanic["age"].between(20, 40)
    & (titanic["pclass"] == 3),
    ["pclass", "age", "fare", "survived"],
]
print("3등석 행 수:", len(male_age_20_40_third))
print("3등석 생존률:", round(male_age_20_40_third["survived"].mean(), 2))
```
